```python
import numpy as np

# ==========================================
# Step 1: Create Matrix A (3x3 Random Integer Matrix)
# ==========================================

A = np.random.randint(1, 10, (3, 3))

print("Matrix A:\n", A)
print("--------------------------------------------------")

# ==========================================
# Step 2: Analyze Matrix A
# ==========================================

# Shape
size_A = A.shape

# Determinant
det_A = np.linalg.det(A)

# Rank
rank_A = np.linalg.matrix_rank(A)

# Eigenvalues
eival_A = np.linalg.eigvals(A)

# Inverse (Safe floating comparison)
if not np.isclose(det_A, 0):
    inv_A = np.linalg.inv(A)
else:
    inv_A = "Matrix A is singular, cannot compute inverse"

print("Shape of Matrix A:", size_A)
print("Determinant of Matrix A:", det_A)
print("Rank of Matrix A:", rank_A)
print("Eigenvalues of Matrix A:", eival_A)
print("Inverse of Matrix A:\n", inv_A)

print("==================================================")

# ==========================================
# Step 3: Value Change Experiment
# Change exactly ONE value in A (+1)
# ==========================================

B = A.copy()

# Change element at position (0,0) by adding +1
B[0, 0] = B[0, 0] + 1

print("Matrix B (Changed element A[0,0] + 1):\n", B)
print("--------------------------------------------------")

# ==========================================
# Step 4: Analyze Matrix B
# ==========================================

# Shape
size_B = B.shape

# Determinant
det_B = np.linalg.det(B)

# Rank
rank_B = np.linalg.matrix_rank(B)

# Eigenvalues
eival_B = np.linalg.eigvals(B)

# Inverse
if not np.isclose(det_B, 0):
    inv_B = np.linalg.inv(B)
else:
    inv_B = "Matrix B is singular, cannot compute inverse"

print("Shape of Matrix B:", size_B)
print("Determinant of Matrix B:", det_B)
print("Rank of Matrix B:", rank_B)
print("Eigenvalues of Matrix B:", eival_B)
print("Inverse of Matrix B:\n", inv_B)
```
