```python
# Necessary Library
import numpy as np
import matplotlib.pyplot as plt

# Ellipse parameters
a = 2
b = 1

# Rectangle limits
x_max = 2
y_max = 1

# Number of random points
N = 10000

# Generate random points
x = np.random.uniform(0, x_max, N)
y = np.random.uniform(0, y_max, N)

# Check if points are inside ellipse
inside = (x**2)/(a**2) + (y**2)/(b**2) <= 1

# Count inside points
inside_points = np.sum(inside)

# Rectangle area
rectangle_area = x_max * y_max

# Estimated area using Monte Carlo
estimated_area = rectangle_area * (inside_points / N)

# True area of quarter ellipse
true_area = np.pi * a * b / 4

# Visualization

plt.figure(figsize=(6,6))

# Points inside ellipse
plt.scatter(x[inside], y[inside], s=5, alpha=0.5, label="Inside Ellipse")

# Points outside ellipse
plt.scatter(x[~inside], y[~inside], s=5, alpha=0.5, label="Outside Ellipse")

plt.title("Monte Carlo Simulation for Quarter Ellipse")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()

plt.show()

# Results

print("Total Points:", N)
print("Points inside ellipse:", inside_points)
print("Estimated Area:", estimated_area)
print("True Area:", true_area)
```
