```python
# Necessary Libraries
import numpy as np
import matplotlib.pyplot as plt

# ===============================
# Part 1 — Data Generation
# ===============================

# 1. Uniform Distribution (Regular Day)
uniform_wait = np.random.uniform(2, 12, 3000)

# 2. Normal Distribution (Busy Day)
normal_wait = np.random.normal(8, 2, 3000)

# 3. Replace negative values with 0
normal_wait = np.where(normal_wait < 0, 0, normal_wait)

# ===============================
# Part 2 — Visualization
# ===============================

plt.figure(figsize=(12, 5))

# Subplot 1 — Uniform
plt.subplot(1, 2, 1)
plt.hist(uniform_wait, bins=30, alpha=0.7)
plt.title("Uniform Distribution (Regular Day)")
plt.xlabel("Wait Time (minutes)")
plt.ylabel("Frequency")

# Subplot 2 — Normal
plt.subplot(1, 2, 2)
plt.hist(normal_wait, bins=30, alpha=0.7)
plt.title("Normal Distribution (Busy Day)")
plt.xlabel("Wait Time (minutes)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()

# ===============================
# Part 3 — Statistical Analysis
# ===============================

print("========== Regular Day (Uniform) ==========")
print("Mean:", np.mean(uniform_wait))
print("Median:", np.median(uniform_wait))
print("Standard Deviation:", np.std(uniform_wait))
print("Maximum Wait Time:", np.max(uniform_wait))

print("\n========== Busy Day (Normal) ==========")
print("Mean:", np.mean(normal_wait))
print("Median:", np.median(normal_wait))
print("Standard Deviation:", np.std(normal_wait))
print("Maximum Wait Time:", np.max(normal_wait))
```
